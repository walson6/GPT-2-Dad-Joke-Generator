# Fine-tune GPT-2

```python
tf.compat.v1.reset_default_graph() # This line is added to reset the graph
sess = gpt2.start_tf_sess()
gpt2.finetune(sess,
              dataset=file_name,
              model_name='124M',
              steps=500,
              restore_from='fresh',
              run_name='dadjokes1',
              print_every=10,
              sample_every=100,
              )
```

# Generate Jokes

# Parameters for generation


Our final parameters were chosen as:

- length = 100 - maximum length of each joke (max is 1023)
- temperature = 1.2 - higher temp means more original jokes and less copies (default is 0.7)
- nsamples = 100 - number of jokes to be generated
- batch_size = 20 - generates multiple samples in parallel to speed up runtime (max is 20)

We also added the below parameters that GPT-2 recognizes to handle the single-line output:

- prefix = "<|startoftext|>" - GPT-2's recognized indicator for starting text
- truncate = "<|endoftext|>" - truncates after each joke
- include_prefix = False - removes prefix from output

- To download a .txt file of our output joke, include the output file as :

- gen_file = 'gpt2_jokes.txt' - can change file name

Additional parameters for output file:

- destination_path = gen_file - output .txt path
- sample_delim = '' - removes "====" from in between each joke

These parameters were chosen to output jokes with the most originality and highest scores.

```python
gen_file = './gpt2_jokes.txt'

gpt2.generate_to_file(sess, run_name = 'dadjokes1',
                      destination_path=gen_file,
                      length=100,
                      temperature=1.2,
                      nsamples=100,
                      batch_size=20,
                      prefix="<|startoftext|>",
                      truncate="<|endoftext|>",
                      include_prefix=False,
                      sample_delim=''
                      )
```

# Output Jokes In Notebook

To output a single joke or group of jokes to the notebook, run the following with the desired number of jokes (nsamples).

If outputting >10 jokes, increase the batch_size to improve runtime.

The parameters length = 100 and temperature = 1.2 can be changed for different joke lengths and joke originality.

Note that a higher temperature will result in more original jokes but also more nonsensical output. A lower temperature will result in less original jokes. The given parameters are optimized for most originality with the highest score.


```python
gpt2.generate(sess, run_name='dadjokes1',
              length=100,
              temperature=1.2,
              nsamples=1,
              batch_size=1,
              prefix="<|startoftext|>",
              truncate="<|endoftext|>",
              include_prefix=False
              )
```

Experimenting with various temperatures, top_k, and top_p values and how these parameters affect the output.

```python
# Temperature controls randomness in predictions. Lower temp is safer, higher temp is more creative
temp_values = [0.7, 1.0, 1.2, 1.5]
for temp in temp_values:
    print(f"temps: {temp}")
    gpt2.generate(sess,
                  run_name='dadjokes1',
                  length=100,
                  temperature=temp,
                  nsamples=3,
                  batch_size=1,
                  prefix="<|startoftext|>",
                  truncate="<|endoftext|>",
                  include_prefix=False)

# Top-k sampling limits the model’s choices to the top k most likely next words according to their probability, and then randomly samples from that list. Lower k is safer, higher k is more diverse
k_values = [0, 40, 80]
for k in k_values:
    print(f"top k: {k}")
    gpt2.generate(sess,
                  run_name='dadjokes1',
                  length=100,
                  temperature=1.0,
                  top_k=k,
                  nsamples=3,
                  batch_size=1,
                  prefix="<|startoftext|>",
                  truncate="<|endoftext|>",
                  include_prefix=False)

# Top_p sampling limits the model's choices to the top p percent of the probability mass instead of top k. Lower p is more focused, higher p is more diverse
p_values = [0.8, 0.9, 1.0]
for p in p_values:
    print(f"top p: {p}")
    gpt2.generate(sess,
                  run_name='dadjokes1',
                  length=100,
                  temperature=1.0,
                  top_p=p,
                  nsamples=3,
                  batch_size=1,
                  prefix="<|startoftext|>",
                  truncate="<|endoftext|>",
                  include_prefix=False)
```

### Temperature:
0.7 produced safer but more repetitive jokes. They made more sense but are less creative. 1.0 gave a good balance of creativity and coherency. 1.2 produced the most creative jokes, but some were incoherent.

### Top-k:
Lower top-k values led to more predictability. It's less creative and more coherent. Higher top-k values led to more creative but sometimes incoherent jokes.

### Top-p:
Since top-p is based on cumulative probability rather than a fixed number, probability is dynamically allocated to include only the most likely words until a certain threshold is reached, which allowed for more adaptive and context-sensitive generation compared to top-k.
